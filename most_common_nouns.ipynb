## extraction of the most common nouns per speech

#### includes
- function to extract nouns from speeches
- iteration over all txt-files saved in folder 'txt_files'

```python
# imports
import spacy
import pathlib
import pandas as pd
from collections import Counter

# load spaCy
nlp = spacy.load("de_core_news_lg")

# txt-files
folder_path = pathlib.Path("txt_files")

# list for all results
results = []

# function to extract nouns
def extract_nouns(text):
    doc = nlp(text)
    
    nouns = [
        token.text
        for token in doc
        if token.pos_ == "NOUN"
    ]
    
    return nouns

# iterate over all txt-files
for file in folder_path.glob("*.txt"):
    
    text = file.read_text(encoding="utf-8")
    
    nouns = extract_nouns(text)

    common_nouns = Counter(nouns).most_common(10)

    speech_id = file.stem
    
    for noun, count in common_nouns:
        results.append({
            "id": speech_id,
            "noun": noun,
            "count": count
        })

# DataFrame
df = pd.DataFrame(results)

# save to csv
df.to_csv("most_common_nouns_speeches.csv",
          sep=";",
          index=False,
          encoding="utf-8")

print(df.head())
```
