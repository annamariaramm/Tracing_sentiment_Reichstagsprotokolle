## preprocessing
#### includes
- lowercasing
- stopword removal
- spelling normalization

```python
#imports
import pandas as pd
from tqdm import tqdm

dataset = pd.read_csv("dataset_speeches.csv", sep=";", dtype="string")
print(dataset.info)

consolidations = pd.read_csv("consolidations.csv", sep=";", dtype="string")
print(consolidations.info)

spelling = pd.read_csv("spelling.csv", sep=";", dtype="string")
print(spelling.info)

stopwords = pd.read_csv("stopwords.csv", sep=";", dtype="string")
print(stopwords.info)

#connecting text columns
dataset["content"] = dataset["text"] + " " + dataset["text_2"] + " " + dataset["text_3"]
print(dataset.info)


#for the entire dataset
for i in tqdm(consolidations.index):
    dataset = dataset.replace(to_replace=consolidations.loc[i].at["letters"], value=consolidations.loc[i].at["replace"], regex=True)

#for the new content column  
dataset["content"] = dataset["content"].str.lower()

for j in tqdm(spelling.index):
    dataset["content"] = dataset["content"].replace(to_replace=" {} ".format(spelling.loc[j].at["word"]), value=" {} ".format(spelling.loc[j].at["replace"]), regex=True)


for k in tqdm(stopwords.index):
    dataset["content"] = dataset["content"].replace(to_replace=" {} ".format(stopwords.loc[k].at["stopwords"]), value=" ", regex=True)

#delete old text
del dataset["text"]
del dataset["text_2"]
del dataset["text_3"]
    
#save to csv
dataset.to_csv("prep_dataset_speeches.csv", sep=";")
print("done")
```
