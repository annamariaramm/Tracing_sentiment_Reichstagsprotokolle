## sentiment analysis
#### the code was originally written by Kara Kuebart (University of Bonn) and was adapted for this study

#### includes
- adding of new columns to count sentiment values and detected words in sentiment lexicon and speeches
- word-by-word check for matches

### sentiment analysis for the context (words surrounding search terms)

```python
#imports
import pandas as pd
from tqdm import tqdm

dataset = pd.read_csv("keyword_contexts.csv", sep=";", dtype="string")
lexicon = pd.read_csv("sentiment_lexicon.csv", sep=";")
lexicon.info()

#sentiment code, add new columns
n = len(dataset.columns)
dataset.insert(loc=n, column="speech_sentiment", value="0", allow_duplicates=True)
n = len(dataset.columns)
dataset.insert(loc=n, column="n_of_lexicon_words", value="0", allow_duplicates=True)
#compare each word with sentiment lexicon
for i in tqdm(dataset.index):
    speech = str(dataset.loc[i].at["context"])
    speech = speech.split(" ")
    speech_sentiment = 0
    n_of_countable_words = 0
    for j in range(len(speech)):
        sf = lexicon.index[lexicon["words"] == speech[j]]
        for m in range(len(sf)):
            n_of_countable_words = n_of_countable_words + 1
            speech_sentiment = speech_sentiment + lexicon.loc[sf[m]].at["value"]
            break
    dataset.at[i, "speech_sentiment"] = speech_sentiment
    dataset.at[i, "n_of_countable_words"] = n_of_countable_words
#save to csv
dataset.to_csv("result_sentiment_contexts.csv", sep=";")
```

### sentiment analysis for the entire speeches
