## extract surrounding terms
#### includes
- iteration over all txt-files
- iteration over all search terms using regular expressions

```python
# imports
import pathlib
import re
import pandas as pd

# txt-files
folder_path = pathlib.Path("txt_files")

# search terms
keywords = [
    "sexu",
    "gleichgeschl",
    "eulenburg",
    "harden",
    "moltke",
    "hirschfeld",
    "komitee",
    "lesb",
    "urani",
    "urni",
    "175",
    "philist",
    "widernat",
    "sodom",
    "kamarill",
    "pädera"
]

# words before and after search term
window = 5

# list for all results
results = []

# iterate over all txt-files
for file in folder_path.glob("*.txt"):

    text = file.read_text(encoding="utf-8")

    # iterate over all search terms
    for keyword in keywords:

        # regex pattern
        pattern = re.compile(
            rf"(?:\b\w+\b\W+){{0,{window}}}"
            rf"\w*{re.escape(keyword)}\w*"
            rf"(?:\W+\b\w+\b){{0,{window}}}",
            re.IGNORECASE
        )

        matches = pattern.finditer(text)

        for match in matches:

            results.append({
                "id": file.stem,
                "keyword": keyword,
                "context": match.group(0)
            })

# DataFrame
df = pd.DataFrame(results)

# save to csv
df.to_csv(
    "keyword_contexts.csv",
    sep=";",
    index=False,
    encoding="utf-8"
)

print(df.head())
```
